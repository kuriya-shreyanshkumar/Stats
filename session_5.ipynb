## Task 2: Random Sampling Simulation

A Python implementation generating 1,000 fake user IDs and drawing a random sample of 50 users using Python's standard `random.sample()` alongside `numpy` and `pandas`.

````python

# Set random seeds for reproducibility
random.seed(42)
np.random.seed(42)

# Generate 1,000 fake user IDs using NumPy and Pandas
user_ids = np.array([f"user{i}" for i in range(1, 1001)])
df_users = pd.DataFrame({"user_id": user_ids})

# 1. Primary Method: Using random.sample() 
sample_users_random = random.sample(list(user_ids), 50)

# 2. Pandas implementation
sample_users_pandas = df_users.sample(n=50, random_state=42)["user_id"].values

print("--- 50 Randomly Selected User IDs (via random.sample) ---")
print(sample_users_random)

print("\nTotal users sampled:", len(sample_users_random))
print("Sample DataFrame shape:", sample_users_pandas.shape)
````

## Task 3: Why Use Sampling for Swiggy Delivery Times Analysis

## Task 4: Identification and Explanation of Bias in Spotify Survey

* **Type of Bias:** **Sampling Bias** (specifically **Selection Bias** / **Undercoverage Bias**).
* **Explanation:** By surveying only users who recently streamed Bollywood songs, the sampling frame systematically excludes users who stream International Pop, Hip-Hop, Classical, Punjabi, Indie, or Regional content. As a result, the sample fails to represent the overall Spotify user base, leading to an overestimation of Bollywood preferences and invalid conclusions regarding the general platform's music consumption patterns.

## Task 5: Sampling Methods for BookMyShow Movie Booking Habits

### 1. Stratified Sampling
* **Description:** The user population is partitioned into mutually exclusive subgroups (strata) based on a key operational variable—such as **City Tier** (Tier 1: Metro, Tier 2: Urban, Tier 3: Semi-Urban) or **Booking Frequency** (Occasional, Regular, High-Frequency).
* **Application on BookMyShow:**
  1. Group the user database by `city_tier`.
  2. Determine the proportion of total users belonging to each tier.
  3. Draw a proportional random sample from each tier using `pandas.DataFrame.groupby().sample()`.
  4. This ensures that smaller demographic segments (e.g., Tier 3 cinema patrons) are accurately represented rather than dominated by high-volume metro bookings.

### 2. Systematic Sampling
* **Description:** Users are selected at equal, predetermined numerical intervals (\(k\)) from a comprehensive, sequentially ordered list.
* **Application on BookMyShow:**
  1. Sort all registered users sequentially by registration timestamp or internal numerical user ID (\(N\) total users).
  2. Set a sampling interval \(k = \frac{N}{n}\) (for example, to sample 1,000 users from a list of 100,000, \(k = 100\)).
  3. Randomly select a starting index between 1 and \(k\), and select every \(100^{\text{th}}\) user thereafter (e.g., user indices 14, 114, 214, 314...).
  4. This provides an even distribution across sign-up cohorts without requiring manual stratification variables.

````python

# Creating a mock dataset representing BookMyShow users
np.random.seed(42)
n_users = 1200

bms_data = pd.DataFrame({
    "user_id": [f"BMS_{i:04d}" for i in range(1, n_users + 1)],
    "city_tier": np.random.choice(["Tier 1", "Tier 2", "Tier 3"], size=n_users, p=[0.5, 0.3, 0.2]),
    "avg_tickets_per_month": np.random.poisson(lam=2, size=n_users)
})

# 1. Stratified Sampling: 10% sample from each city tier
stratified_sample = (
    bms_data.groupby("city_tier", group_keys=False)
    .sample(frac=0.10, random_state=42)
)

# 2. Systematic Sampling: Selecting every k-th user
k = 10
start_idx = np.random.randint(0, k)
systematic_sample = bms_data.iloc[start_idx::k]

print("Stratified Sample Breakdown by Tier:")
print(stratified_sample["city_tier"].value_counts())

print(f"\nSystematic Sample Size (every {k}th record from index {start_idx}):", len(systematic_sample))
````
