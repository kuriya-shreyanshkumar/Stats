## Task 1: Basic Probability Calculation Function
Write a Python function `calculate_probability(event_count, total_outcomes)` that returns the probability of an event, and use it to find the probability of getting 'Heads' when tossing a coin.

```python
def calculate_probability(event_count, total_outcomes):
    """
    Calculates the theoretical probability of an event.
    Formula: P(E) = Number of favorable outcomes / Total number of possible outcomes
    """
    if total_outcomes == 0:
        raise ValueError("Total outcomes cannot be zero.")
    return event_count / total_outcomes

# Tossing a fair coin: 1 'Head' outcome out of 2 total outcomes ('Heads', 'Tails')
event_count_heads = 1
total_outcomes_coin = 2

prob_heads = calculate_probability(event_count_heads, total_outcomes_coin)
print(f"Probability of getting 'Heads': {prob_heads}")
```

## Task 2: Dice Roll Simulation (Experimental Probability)
Simulate rolling a six-sided dice 100 times using Python's `random` module, count how many times a '6' appears, and calculate the experimental probability of getting a '6'.

```python


# Simulation parameters
trials = 100
count_six = 0

# Simulating 100 dice rolls
for _ in range(trials):
    roll = random.randint(1, 6)
    if roll == 6:
        count_six += 1

# Experimental probability = count of target outcome / total trials
experimental_prob_six = count_six / trials

print(f"Total rolls: {trials}")
print(f"Number of times '6' appeared: {count_six}")
print(f"Experimental probability of rolling a '6': {experimental_prob_six}")
```

## Task 3: Probability from Categorical Data (Spotify Genres)
Given an array of genres from your last 20 Spotify songs, calculate the probability of randomly picking a 'pop' song from your recent history using `numpy` and `pandas`.

```python

# Array of genres from the last 20 Spotify songs
spotify_genres = np.array([
    'pop', 'rock', 'pop', 'hiphop', 'jazz',
    'pop', 'indie', 'rock', 'pop', 'classical',
    'hiphop', 'pop', 'electronic', 'rock', 'pop',
    'pop', 'jazz', 'r&b', 'pop', 'rock'
])

# Load into a pandas Series for frequency calculation
genre_series = pd.Series(spotify_genres)

# Count occurrences of 'pop' and total songs
pop_count = (genre_series == 'pop').sum()
total_songs = len(genre_series)

# Probability calculation
prob_pop = pop_count / total_songs

print(f"Total songs: {total_songs}")
print(f"Number of 'pop' songs: {pop_count}")
print(f"Probability of picking a 'pop' song: {prob_pop}")
```

## Task 4: Intersection Over Total Unique Users (Zomato Orders)
Write a function that, given two arrays — one for users who ordered food on Zomato and one for users who ordered dessert — calculates the probability that a randomly selected user ordered both food and dessert (intersection over total unique users).

```python


def prob_food_and_dessert(food_users, dessert_users):
    """
    Calculates P(Food and Dessert) = Count(Food ∩ Dessert) / Count(Food ∪ Dessert)
    """
    # Convert inputs to unique sets via numpy
    unique_food = np.unique(food_users)
    unique_dessert = np.unique(dessert_users)
    
    # Intersection: users who ordered both
    intersection_users = np.intersect1d(unique_food, unique_dessert)
    
    # Union: total unique users across both groups
    union_users = np.union1d(unique_food, unique_dessert)
    
    if len(union_users) == 0:
        return 0.0
        
    return len(intersection_users) / len(union_users)

# Sample user data arrays
food_orders = np.array(['user_1', 'user_2', 'user_3', 'user_4', 'user_5', 'user_6'])
dessert_orders = np.array(['user_3', 'user_4', 'user_7', 'user_8'])

prob_both = prob_food_and_dessert(food_orders, dessert_orders)

print(f"Probability that a user ordered both food and dessert: {prob_both:.4f}")
```

## Task 5: Recommendation System via Conditional Probability (Flipkart)

Conditional probability measures the likelihood of an event occurring given that another event has already occurred:

$$P(\text{Headphones} \mid \text{Phone}) = \frac{\text{Count}(\text{Bought Both Phone and Headphones})}{\text{Count}(\text{Bought Phone})}$$

```python


# Sample purchase records for 10 users
transactions = pd.DataFrame({
    'user_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'bought_phone': [True, True, True, True, True, False, False, True, False, True],
    'bought_headphones': [True, False, True, True, False, True, False, True, False, False]
})

# Filter users who bought a phone
phone_buyers = transactions[transactions['bought_phone'] == True]
count_phone = len(phone_buyers)

# Filter users who bought both a phone and headphones
both_buyers = transactions[(transactions['bought_phone'] == True) & (transactions['bought_headphones'] == True)]
count_both = len(both_buyers)

# Calculate conditional probability: P(Headphones | Phone)
prob_headphones_given_phone = count_both / count_phone if count_phone > 0 else 0.0

print(f"Total users who bought a phone: {count_phone}")
print(f"Users who bought both phone & headphones: {count_both}")
print(f"P(Headphones | Phone): {prob_headphones_given_phone:.2f}")
```
