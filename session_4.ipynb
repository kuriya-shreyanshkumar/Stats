```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Task 1: Swiggy Daily Order Counts

**Problem:** Given the daily order counts for a Swiggy delivery partner over 10 days: `[12, 15, 17, 14, 13, 16, 200, 18, 14, 15]`, plot a simple graph and describe whether the data is left skewed, right skewed, or symmetrical. Explain your reasoning based on the shape.

```python

orders = pd.Series([12, 15, 17, 14, 13, 16, 200, 18, 14, 15])

plt.figure(figsize=(8, 4))
plt.hist(orders, bins=15, edgecolor='black', color='skyblue')
plt.title('Distribution of Daily Swiggy Orders')
plt.xlabel('Order Count')
plt.ylabel('Frequency (Days)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# Verification using summary statistics
print(f"Mean: {orders.mean():.2f}")
print(f"Median: {orders.median():.2f}")
```

### Task 1 Answer:
* **Distribution Type:** **Right Skewed** (Positively Skewed).
* **Reasoning:** Most values lie clustered between 12 and 18, while the extreme outlier of `200` stretches the tail far out to the right side of the graph. Because the tail extends towards the higher values (and Mean: 31.90 > Median: 15.00), the distribution is right-skewed.

## Task 2: Instagram Influencer Follower Counts

**Problem:** Take the following list of Instagram influencer follower counts (in thousands): `[5, 7, 8, 8, 9, 10, 12, 15, 95]`. Calculate the mean and median, then state which is higher and what that tells you about the skewness of the data.

```python

followers = pd.Series([5, 7, 8, 8, 9, 10, 12, 15, 95])

mean_val = followers.mean()
median_val = followers.median()

print(f"Mean: {mean_val:.2f}k")
print(f"Median: {median_val:.2f}k")
```

### Task 2 Answer:
* **Calculated Values:**
  * **Mean:** 18.78 (thousand followers)
  * **Median:** 9.00 (thousand followers)
* **Comparison & Skewness:** 
  * The **mean is higher than the median** (\(18.78 > 9.00\)).
  * This indicates that the data is **right skewed** (positively skewed) because the extreme high value (`95`) pulls the mean up and away from the center of the distribution.

## Task 3: Real-World Example of Right-Skewed Data

**Problem:** Find a real-world example of right-skewed data from any app you use (like Zomato order amounts, Paytm wallet balances, or YouTube video views). Briefly describe the dataset and explain why you think it is right skewed.

### Task 3 Answer:

* **Selected App & Dataset:** **YouTube Video Views**
* **Description:** A dataset containing the total view count for all videos uploaded across YouTube.
* **Why it is Right Skewed:**
  * The majority of videos uploaded by typical users or smaller creators receive a modest number of views (ranging from tens to a few thousand views).
  * A very small fraction of viral videos or music videos accumulate hundreds of millions or billions of views.
  * Because the lower bound is zero and a small cluster of massive outlier videos stretches the tail far toward higher numbers, the mean view count is pulled significantly to the right of the median.

```python
# Task 3: Demonstration using simulated YouTube view counts
np.random.seed(42)
# Simulating a log-normal distribution to represent typical YouTube view counts
simulated_views = pd.Series(np.random.lognormal(mean=7, sigma=1.5, size=1000))

print("YouTube Views Simulation:")
print(f"Median Views: {simulated_views.median():.0f}")
print(f"Mean Views:   {simulated_views.mean():.0f}")
print(f"Mean > Median confirms Right Skewness: {simulated_views.mean() > simulated_views.median()}")
```

## Task 4: Flipkart Product Ratings

**Problem:** Given this dataset of Flipkart product ratings: `[3, 3, 4, 4, 4, 5, 5, 5, 5, 5]`, draw a rough graph and state if the distribution is symmetrical, left skewed, or right skewed. Justify your answer in one line.

```python
# Task 4: Ratings distribution plot
ratings = pd.Series([3, 3, 4, 4, 4, 5, 5, 5, 5, 5])

rating_counts = ratings.value_counts().sort_index()

plt.figure(figsize=(6, 4))
plt.bar(rating_counts.index, rating_counts.values, edgecolor='black', color='salmon', width=0.5)
plt.title('Flipkart Product Ratings Count')
plt.xlabel('Rating (Stars)')
plt.ylabel('Count')
plt.xticks([1, 2, 3, 4, 5])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

print(f"Mean: {ratings.mean():.2f}")
print(f"Median: {ratings.median():.2f}")
```

### Task 4 Answer:
* **Distribution Type:** **Left Skewed** (Negatively Skewed).
* **Justification:** The data is left skewed because the majority of ratings are clustered at the higher end (4 and 5), while the tail of lower values (3) extends to the left (Mean: 4.30 < Median: 4.50).
