
## Task 1: Calculate Variance by Hand (Step-by-Step)

### Given Information
- **Dataset (Daily Steps):** \(X = [4500, 7000, 5000, 8000, 4000, 9000, 6000]\)
- **Number of observations \(N\):** $7$



---

### Step 1: Write Down the Variance Formula

For a complete set of observations (population):

$$\sigma^2 = \frac{\sum_{i=1}^{N} (x_i - \mu)^2}{N}$$

Where:

* $x_i$ = each individual value
* $\mu$ = mean (average) of the values
* $N$ = total count of values
* $\sum$ = summation symbol (add all values together)

---

### Step 2: Calculate the Mean ($\mu$)

1. **Sum all the step counts:**

$$\sum x_i = 4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000 = 43,500$$

2. **Divide by the total count ($N = 7$):**

$$\mu = \frac{43,500}{7} \approx 6214.29$$

---

### Step 3: Calculate the Deviation for Each Value $(x_i - \mu)$

Subtract the mean ($\mu = 6214.29$) from each individual step count:

* **Day 1 ($x_1$):** $4500 - 6214.29 = -1714.29$
* **Day 2 ($x_2$):** $7000 - 6214.29 = +785.71$
* **Day 3 ($x_3$):** $5000 - 6214.29 = -1214.29$
* **Day 4 ($x_4$):** $8000 - 6214.29 = +1785.71$
* **Day 5 ($x_5$):** $4000 - 6214.29 = -2214.29$
* **Day 6 ($x_6$):** $9000 - 6214.29 = +2785.71$
* **Day 7 ($x_7$):** $6000 - 6214.29 = -214.29$

*(Check: The sum of deviations $\sum (x_i - \mu) \approx 0$)*

---

### Step 4: Square Each Deviation $(x_i - \mu)^2$

Squaring eliminates negative values:

* **Day 1:** $(-1714.29)^2 = 2,938,789.80$
* **Day 2:** $(785.71)^2 = 617,340.20$
* **Day 3:** $(-1214.29)^2 = 1,474,500.20$
* **Day 4:** $(1785.71)^2 = 3,188,760.20$
* **Day 5:** $(-2214.29)^2 = 4,903,080.20$
* **Day 6:** $(2785.71)^2 = 7,760,180.20$
* **Day 7:** $(-214.29)^2 = 45,920.20$

---

### Step 5: Summary Table of Calculations

| Day ($i$) | Steps ($x_i$) | Mean ($\mu$) | Deviation $(x_i - \mu)$ | Squared Deviation $(x_i - \mu)^2$ |
| --- | --- | --- | --- | --- |
| 1 | 4500 | 6214.29 | $-1714.29$ | $2,938,789.80$ |
| 2 | 7000 | 6214.29 | $+785.71$ | $617,340.20$ |
| 3 | 5000 | 6214.29 | $-1214.29$ | $1,474,500.20$ |
| 4 | 8000 | 6214.29 | $+1785.71$ | $3,188,760.20$ |
| 5 | 4000 | 6214.29 | $-2214.29$ | $4,903,080.20$ |
| 6 | 9000 | 6214.29 | $+2785.71$ | $7,760,180.20$ |
| 7 | 6000 | 6214.29 | $-214.29$ | $45,920.20$ |
| **Total** | **43,500** | — | **0.00** | **$\sum (x_i - \mu)^2 = 20,928,571.43$** |

---

### Step 6: Sum the Squared Deviations ($\sum (x_i - \mu)^2$)

$$\sum_{i=1}^{7} (x_i - \mu)^2 = 2,938,789.80 + 617,340.20 + 1,474,500.20 + 3,188,760.20 + 4,903,080.20 + 7,760,180.20 + 45,920.20$$

$$\sum_{i=1}^{7} (x_i - \mu)^2 = 20,928,571.43$$

---

### Step 7: Compute the Final Variance

 Population Variance ($\sigma^2$) — dividing by $N = 7$:

$$\sigma^2 = \frac{20,928,571.43}{7} \approx 2,989,795.92 \text{ steps}^2$$



**Final Result:**

* **Population Variance ($\sigma^2$):** **`2,989,795.92`**


```

```

````python

steps = np.array([4500, 7000, 5000, 8000, 4000, 9000, 6000])

df_steps = pd.DataFrame({"Steps": steps})
mean_val = df_steps["Steps"].mean()
df_steps["Deviation"] = df_steps["Steps"] - mean_val
df_steps["Squared_Deviation"] = df_steps["Deviation"] ** 2

pop_variance = np.var(steps, ddof=0)


print(df_steps)
print(f"\nMean Steps: {mean_val:.2f}")
print(f"Population Variance (N=7): {pop_variance:.2f}")
````

## Task 2: Function `calculate_standard_deviation(scores)`

Write a Python function called `calculate_standard_deviation(scores)` that takes a list of exam scores and returns the standard deviation rounded to two decimal places.

````python
def calculate_standard_deviation(scores):
    scores_array = np.array(scores)
    std_dev = np.std(scores_array)
    return round(float(std_dev), 2)

# Function Call / Test
exam_scores = [78, 85, 92, 65, 88, 72, 90]
result = calculate_standard_deviation(exam_scores)
print("Exam Scores:", exam_scores)
print("Standard Deviation:", result)
````

## Task 3: Zomato Spending Comparison

- **Friend A:** `[200, 200, 200, 200, 200]`
- **Friend B:** `[100, 300, 150, 400, 50]`

````python
friend_a = np.array([200, 200, 200, 200, 200])
friend_b = np.array([100, 300, 150, 400, 50])

std_friend_a = np.std(friend_a)
std_friend_b = np.std(friend_b)

df_spending = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Friend_A": friend_a,
    "Friend_B": friend_b
})

print(df_spending)
print(f"\nFriend A Mean: {np.mean(friend_a):.2f}, Standard Deviation: {std_friend_a:.2f}")
print(f"Friend B Mean: {np.mean(friend_b):.2f}, Standard Deviation: {std_friend_b:.2f}")
````

### Interpretation: Consistency Analysis
Friend A has significantly more consistent spending than Friend B because Friend A's standard deviation is **0.00**, meaning there is zero variation from the daily average of ₹200. In contrast, Friend B has a high standard deviation of **130.38**, showing wide fluctuations in spending from as low as ₹50 to as high as ₹400 across the 5 days.

## Task 4: Startup Monthly Salary Analysis

**Monthly Salaries of 6 Employees:** `[28000, 29000, 27000, 60000, 26500, 27500]`

````python
salaries = np.array([28000, 29000, 27000, 60000, 26500, 27500])
salary_series = pd.Series(salaries, name="Monthly_Salary")

# Population Variance & Standard Deviation
pop_var_salary = np.var(salaries, ddof=0)
pop_std_salary = np.std(salaries, ddof=0)


print(salary_series.describe())
print(f"\nPopulation Variance: {pop_var_salary:.2f}")
print(f"Population Standard Deviation: {pop_std_salary:.2f}")
````

### Interpretation: High vs Low Spread
The data exhibits a high spread with a population variance of **146,416,666.67** and standard deviation of **12,100.28**. This high spread indicates substantial pay disparity within the startup: five employees have very low variation, earning between ₹26,500 and ₹29,000, while a single outlier of ₹60,000 skews the mean and inflates the overall team deviation.
