## Task 1: 2x2 Matrix Multiplication (Instagram Metrics)
Multiply two 2x2 matrices representing likes and comments on two Instagram posts across two days.

```python

# Matrix A: Day 1 metrics [Likes, Comments] for [Post 1, Post 2]
matrix_a = np.array([[120, 15], [250, 40]])

# Matrix B: Day 2 transition weights/metrics [Likes, Comments]
matrix_b = np.array([[1.1, 0.2], [0.8, 1.5]])

# Matrix multiplication using NumPy
result_matrix = np.matmul(matrix_a, matrix_b)

# Displaying inputs and output using Pandas DataFrames
df_a = pd.DataFrame(
    matrix_a,
    columns=['Likes', 'Comments'],
    index=['Day 1 - Post 1', 'Day 1 - Post 2'],
)
df_b = pd.DataFrame(
    matrix_b,
    columns=['Factor 1', 'Factor 2'],
    index=['Weight Likes', 'Weight Comments'],
)
df_result = pd.DataFrame(
    result_matrix,
    columns=['Metric 1', 'Metric 2'],
    index=['Post 1 Result', 'Post 2 Result'],
)

print('--- Matrix A ---')
display(df_a)

print('\n--- Matrix B ---')
display(df_b)

print('\n--- Resulting Matrix (A x B) ---')
display(df_result)
```

## Task 2: Linear Regression Prediction Function
Create a function `linear_regression_predict(x, m, c)` that calculates predicted output \(y = mx + c\).

```python

def linear_regression_predict(x, m, c):
  """Predicts the output y for a linear regression model: y = mx + c.

  Parameters:
  x (int, float, list, np.ndarray): Input feature value(s).
  m (float): Slope of the line.
  c (float): Y-intercept.

  Returns:
  np.ndarray or float: Predicted value(s) y.
  """
  x_arr = np.array(x)
  y = (m * x_arr) + c
  return y
```

## Task 3: Zomato Delivery Time Prediction
Predict delivery time \(y) for distances 2 km, 5 km, and 8 km using the equation \(y = 4x + 10\).

```python
# Given parameters
distances = np.array([2, 5, 8])  # in km (x)
slope = 4  # m (minutes per km)
intercept = 10  # c (base prep/packaging time in minutes)

# Calculate predicted delivery times
predicted_times = linear_regression_predict(distances, slope, intercept)

# Structure and display results
zomato_df = pd.DataFrame({
    'Distance (km)': distances,
    'Equation': [f'4({d}) + 10' for d in distances],
    'Predicted Delivery Time (mins)': predicted_times,
})

zomato_df
```

## Task 4: Slope and Intercept in Flipkart Price Prediction

* **Slope (\(m\)):** 
In Flipkart's product price prediction model, the slope represents the rate of change in product price for every one-unit increase in customer rating. For example, a positive slope indicates how many extra rupees a seller can charge as the product rating climbs from 3 stars to 4 stars. It captures the direct sensitivity and relationship between buyer satisfaction and market pricing.

* **Intercept (\(c\)):** 
The intercept represents the baseline estimated price of a product when its rating is zero. In a business context, this accounts for the inherent floor value—such as basic manufacturing, packaging, and logistical costs—independent of customer feedback. It ensures that an unrated or zero-rated product still retains a baseline monetary benchmark.

## Task 5: Matrix Multiplication in Spotify's Recommendation System

In Spotify's recommendation engine, collaborative filtering breaks down massive listening history into two lower-dimensional matrices: a User-Feature matrix and a Song-Feature matrix. Matrix multiplication is used to take the dot product between a user’s latent preference vector and a song’s acoustic/genre factor vector. This rapid vector-matrix operation outputs predicted affinity scores across millions of tracks simultaneously. In model training, matrix multiplication drives backpropagation and alternating least squares (ALS) to continuously minimize the difference between predicted and actual listening behavior.
