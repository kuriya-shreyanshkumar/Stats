```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Configure plot styling
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
```

## Task 1: Instagram Usage vs. Number of Posts

Create two Python lists representing 7 days of daily Instagram hours and daily posts made. Plot them as a scatter plot and visually describe the relationship.

```python
# Data for 7 days of a week
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
instagram_hours = [1.5, 2.0, 1.0, 3.5, 2.5, 4.0, 4.5]
instagram_posts = [1, 2, 0, 3, 2, 4, 5]

# Create DataFrame
df_insta = pd.DataFrame({
    'Day': days,
    'Hours_Spent': instagram_hours,
    'Posts_Made': instagram_posts
})

print("--- Instagram Data ---")
print(df_insta)

# Plot scatter plot
plt.figure(figsize=(7, 5))
plt.scatter(df_insta['Hours_Spent'], df_insta['Posts_Made'], color='purple', s=80, edgecolors='black')
plt.title('Instagram Usage vs. Daily Posts', fontsize=13, fontweight='bold')
plt.xlabel('Hours Spent on Instagram (per day)', fontsize=11)
plt.ylabel('Number of Posts Made (per day)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Task 1: Visual Description

The scatter plot displays an upward trend moving from left to right. As the number of hours spent on Instagram increases, the number of posts made also generally increases.

This indicates a **positive correlation** between time spent on the platform and posting frequency.

## Task 2: Monthly Zomato vs. Swiggy Orders

Given the monthly order count of 6 friends across Zomato and Swiggy, calculate the correlation coefficient between the two lists using `numpy.corrcoef()`, print the result, and explain if the correlation is positive, negative, or zero.

```python
# Monthly order data for 6 friends
friends = ['Friend 1', 'Friend 2', 'Friend 3', 'Friend 4', 'Friend 5', 'Friend 6']
zomato_orders = np.array([14, 6, 20, 8, 25, 16])
swiggy_orders = np.array([12, 5, 18, 9, 22, 15])

# DataFrame representation
df_orders = pd.DataFrame({
    'Friend': friends,
    'Zomato_Orders': zomato_orders,
    'Swiggy_Orders': swiggy_orders
})
print("--- Food Delivery Orders ---")
print(df_orders)

# Calculate Pearson correlation coefficient using NumPy
corr_matrix = np.corrcoef(zomato_orders, swiggy_orders)
r_value = corr_matrix[0, 1]

print("\n--- Correlation Matrix ---")
print(corr_matrix)
print(f"\nCorrelation Coefficient (r): {r_value:.4f}")
```

### Task 2: Interpretation

* **Correlation Coefficient (r):** ~0.99
* **Result:** **Strong Positive Correlation**
* **Explanation:** Because the correlation coefficient is close to +1.0, there is a very strong positive correlation between Zomato and Swiggy orders. Friends who order food frequently online tend to use both platforms actively, whereas friends who rarely order food place fewer orders across both apps.

## Task 4: Spotify Time vs. Test Marks (Strong Negative Correlation)

Dataset of 10 pairs showing a strong negative correlation between daily hours spent on Spotify and marks scored in a test, visualized as a scatter plot.

```python
# 10 pairs showing strong negative correlation
spotify_hours = np.array([0.5, 1.0, 1.5, 2.0, 2.8, 3.2, 3.8, 4.2, 5.0, 5.5])
test_marks = np.array([96, 91, 88, 82, 76, 73, 67, 62, 55, 48])

# Create DataFrame
df_spotify = pd.DataFrame({
    'Daily_Spotify_Hours': spotify_hours,
    'Test_Marks': test_marks
})

print("--- Spotify vs Test Marks Dataset ---")
print(df_spotify)

# Calculate correlation using NumPy
r_spotify = np.corrcoef(spotify_hours, test_marks)[0, 1]
print(f"\nCorrelation Coefficient (r): {r_spotify:.4f}")

# Plot scatter plot
plt.figure(figsize=(7, 5))
plt.scatter(df_spotify['Daily_Spotify_Hours'], df_spotify['Test_Marks'], color='crimson', s=80, edgecolors='black')

# Line of best fit using numpy polyfit
slope, intercept = np.polyfit(spotify_hours, test_marks, 1)
plt.plot(spotify_hours, slope * spotify_hours + intercept, color='black', linestyle='--', label=f'Fit line (r = {r_spotify:.2f})')

plt.title('Daily Spotify Hours vs. Test Marks', fontsize=13, fontweight='bold')
plt.xlabel('Daily Spotify Hours', fontsize=11)
plt.ylabel('Test Marks (out of 100)', fontsize=11)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

### Task 4: Observation

* The data points clearly slope downward from left to right.
* The correlation coefficient is \(r \approx -0.99\), confirming a **strong negative correlation** where higher daily time spent on Spotify is associated with lower test scores in this dataset.
