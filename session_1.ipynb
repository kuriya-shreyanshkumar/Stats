## Task 2: Music App Data Classification

**Sample Data:** `['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']`

**Classification:** Categorical Data (Nominal)

**Reasoning:**
The given data consists of qualitative labels representing music genres rather than measurable quantities. Because there is no inherent ranking, order, or mathematical relationship between 'Pop', 'Rock', 'Jazz', and 'Hip-Hop' (one genre is not numerically greater or smaller than another), it is classified specifically as nominal categorical data.

```python

genres = ['Pop', 'Rock', 'Jazz', 'Hip-Hop', 'Pop', 'Rock', 'Jazz', 'Pop']

# Frequency calculation (standard approach for categorical data)
genre_counts = Counter(genres)

print("Unique Categories:", set(genres))
print("Category Frequencies:", dict(genre_counts))
print("Data Type Classification: Categorical (Nominal)")
```

## Task 3: Improving an App Feature Using Statistics

**Feature Selected:** Restaurant Rating System on Zomato (e.g., 4.2 / 5.0)

**How Statistics Improves This Feature:**
Currently, a simple arithmetic average can be misleading; a restaurant with two 5-star ratings displays a perfect 5.0, while an established restaurant with thousands of reviews might sit at 4.4 despite higher reliability. By applying statistical techniques such as **weighted Bayesian means** (weighting ratings by total review count) and analyzing **variance/interquartile range** to detect and remove outlier or spam reviews, Zomato can compute a normalized confidence score that gives users a far more accurate representation of restaurant quality.

## Task 4: Statistical Terms for Playlist Recommendations (Spotify-like App)

1. **Mode**
   - *Definition:* The value that appears most frequently in a dataset.
   - *Example:* Calculating the most repeated music genre in a user's weekly play history to choose the primary seed genre for their "Daily Mix" playlist.

2. **Mean**
   - *Definition:* The arithmetic average of a set of numerical values.
   - *Example:* Computing a user's average track duration (e.g., 3.5 minutes) to construct a playlist that precisely fills their standard 30-minute commute.

3. **Variance**
   - *Definition:* A measure of how spread out numbers are from their average value.
   - *Example:* Evaluating the variance in tempo (Beats Per Minute) across saved workout songs to ensure recommended running tracks stay consistently paced rather than jumping between slow and fast tempos.

```python

track_tempos = [120, 128, 128, 130, 125, 128, 122, 135]

mode_tempo = stats.mode(track_tempos, keepdims=True)[0][0]
mean_tempo = np.mean(track_tempos)
variance_tempo = np.var(track_tempos)

print(f"Dataset (BPM): {track_tempos}")
print(f"Mode Tempo: {mode_tempo} BPM  -> Identifies the dominant tempo")
print(f"Mean Tempo: {mean_tempo:.2f} BPM  -> Average tempo baseline")
print(f"Tempo Variance: {variance_tempo:.2f}     -> Spread/diversity of tempo across songs")
```
