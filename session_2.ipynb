## Task 1: Daily Spotify Song Play Counts (Mean)
Create a Python script / code that takes a list of daily Spotify song play counts and calculates the mean (average) number of plays.

```python

play_counts = [120, 135, 150, 200, 120, 90, 200]

# Calculate mean using NumPy and Pandas
mean_np = np.mean(play_counts)
mean_pd = pd.Series(play_counts).mean()

print(f"Daily Play Counts: {play_counts}")
print(f"Mean Plays (NumPy):  {mean_np:.2f}")
print(f"Mean Plays (Pandas): {mean_pd:.2f}")
```

## Task 2: Zomato Order Delivery Times (Median)
Find the median delivery time (in minutes) from the given array of Zomato orders and print it.

```python

delivery_times = [30, 25, 40, 35, 30, 45, 30]

# Calculate median using NumPy and Pandas
median_np = np.median(delivery_times)
median_pd = pd.Series(delivery_times).median()

print(f"Delivery Times: {delivery_times}")
print(f"Median Delivery Time (NumPy):  {median_np} minutes")
print(f"Median Delivery Time (Pandas): {median_pd} minutes")
```

## Task 3: Flipkart Product Ratings (Mode)
Write a function `most_common_rating(ratings)` that takes a list of Flipkart product ratings and returns the mode (most frequent rating).

```python

def most_common_rating(ratings):
    # Using pandas Series mode
    rating_series = pd.Series(ratings)
    mode_values = rating_series.mode()
    return mode_values.iloc[0]

# Given ratings list
ratings = [5, 4, 4, 3, 5, 4, 2, 4]
result = most_common_rating(ratings)

print(f"Flipkart Ratings: {ratings}")
print(f"Most Common Rating (Mode): {result}")
```

## Task 4: Compare YouTube Video Views (Outlier Impact)
Compare the mean, median, and mode for three YouTube channels and determine which channel's data is most affected by outliers.

```python

channels_data = {
    "Channel A": [100, 120, 110, 105, 115],
    "Channel B": [200, 250, 220, 210, 240],
    "Channel C": [100, 120, 110, 105, 5000]  # Extreme outlier present
}

results = []

for channel, views in channels_data.items():
    s = pd.Series(views)
    mean_val = s.mean()
    median_val = s.median()
    mode_val = s.mode().tolist()
    difference = abs(mean_val - median_val)
    
    results.append({
        "Channel": channel,
        "Mean": mean_val,
        "Median": median_val,
        "Mode": mode_val,
        "|Mean - Median|": difference
    })

comparison_df = pd.DataFrame(results)
print(comparison_df.to_string(index=False))

# Identify channel most affected by outliers
outlier_channel = comparison_df.loc[comparison_df["|Mean - Median|"].idxmax(), "Channel"]
print(f"\nDecision: {outlier_channel} is most affected by outliers because the mean is heavily skewed compared to the median.")
```

## Task 5: Real Dataset Analysis with Pandas
Load a real dataset using pandas and display the mean, median, and mode for one numeric column.

```python
 
# Using a publicly hosted CSV dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

# Selected numeric column: 'Fare'
column = "Fare"

# Calculations using pandas and numpy
col_mean = df[column].mean()
col_median = df[column].median()
col_mode = df[column].mode().iloc[0]

print(f"Dataset Shape: {df.shape}")
print(f"Selected Column: '{column}'\n")
print(f"Mean {column}:   {col_mean:.2f}")
print(f"Median {column}: {col_median:.2f}")
print(f"Mode {column}:   {col_mode:.2f}")
```
